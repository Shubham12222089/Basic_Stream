```python
import pandas as pd

# Define functions to calculate match score and shortlist candidates
def calculate_match_score(candidate_skills, job_skills):
    """Calculates match score between candidate skills and job skills"""
    common_skills = set(candidate_skills).intersection(job_skills)
    match_score = len(common_skills) / len(job_skills) * 100  # Express as percentage
    return match_score

def update_shortlisted(match_id, shortlisted):
    """Simulates updating the shortlist status in a database or list"""
    print(f"Candidate with match ID {match_id} {'shortlisted' if shortlisted else 'removed from shortlist'}.")

def generate_match_report(job_id, shortlisted_candidates, job_matches):
    """Generate a report for shortlisted candidates"""
    # Get the job details using job_id
    job = job_matches[job_matches["id"] == job_id].iloc[0]
    
    report_data = {"job": job, "candidates": []}
    
    for candidate_id in shortlisted_candidates:
        candidate = job_matches[job_matches["candidate_id"] == candidate_id].iloc[0]
        candidate_data = {
            "name": candidate["name"],
            "email": candidate["email"],
            "match_score": candidate["match_score"],
            "education": candidate["education"].split(", "),
            "experience": candidate["experience"].split(", "),
            "matching_skills": set(candidate["skills"].split(", ")).intersection(job["skills"].split(", ")),
            "missing_skills": set(job["skills"].split(", ")) - set(candidate["skills"].split(", "))
        }
        report_data["candidates"].append(candidate_data)
    return report_data


# Sample job description and candidates data
job_matches = pd.DataFrame([
    {"id": 1, "candidate_id": 1, "name": "Alice", "email": "alice@example.com", "skills": "python, java, databases", "education": "Bachelor's in CS", "experience": "2 years in software development", "match_score": 80},
    {"id": 1, "candidate_id": 2, "name": "Bob", "email": "bob@example.com", "skills": "python, java, web development", "education": "Master's in CS", "experience": "3 years in full-stack development", "match_score": 70},
    {"id": 1, "candidate_id": 3, "name": "Charlie", "email": "charlie@example.com", "skills": "python, java, databases, web development", "education": "Bachelor's in CS", "experience": "1 year in software engineering", "match_score": 90}
])

# Define job description (as a dictionary)
job_description = {"id": 1, "title": "Software Engineer", "skills": "python, java, databases, web development", "experience": "Not specified", "qualifications": "Bachelor's in Computer Science"}

# Simulate selecting candidates with match score > 70%
threshold = 70
selected_candidates = job_matches[job_matches["match_score"] > threshold]["candidate_id"].tolist()

# Shortlisting selected candidates
for candidate_id in selected_candidates:
    match_id = job_matches[job_matches["candidate_id"] == candidate_id]["id"].iloc[0]
    update_shortlisted(match_id, True)

# Simulate generating a report for shortlisted candidates
shortlisted_candidates = job_matches[job_matches["match_score"] > threshold]["candidate_id"].tolist()
report_data = generate_match_report(job_description["id"], shortlisted_candidates, job_matches)

# Display report data
if report_data:
    print(f"Shortlist Report for {job_description['title']}")
    print("\n### Job Requirements")
    print(f"**Required Skills:** {', '.join(job_description['skills'].split(', '))}")
    print(f"**Experience:** {job_description['experience']}")
    print(f"**Qualifications:** {job_description['qualifications']}")
    
    print("\n### Shortlisted Candidates")
    for candidate in report_data['candidates']:
        print(f"\n{candidate['name']} ({candidate['match_score']}% match)")
        print(f"**Email:** {candidate['email']}")
        print(f"**Match Score:** {candidate['match_score']}%")
        
        print("\n**Education:**")
        for edu in candidate['education']:
            print(f"• {edu}")
        
        print("\n**Work Experience:**")
        for exp in candidate['experience']:
            print(f"• {exp}")
        
        print("\n**Skills Match:**")
        print(f"Matching Skills: {', '.join(candidate['matching_skills'])}")
        print(f"Missing Skills: {', '.join(candidate['missing_skills'])}")
```
